##Type conversions

###Genres

```python
df['genres'][0]
```

```python
import ast
def converter(x):
  L = []
  for i in ast.literal_eval(x):
    L.append(i['name'])
  return L
```

```python
df['genres'] = df['genres'].apply(converter)
df['genres'].sample(5)
```

###keywords

```python
df['keywords'][0]
```

```python
df['keywords'] = df['keywords'].apply(converter)
df['keywords'].sample(5)
```

```python
df['keywords'][0]
```

###Cast (TOP 3)

```python
df['cast'][0]
```

```python
def fetch_actors(x):
  names = []
  count = 0
  for i in ast.literal_eval(x):
    if count < 3:
      names.append(i['name'])
      count+=1
  return names
```

```python
df['cast'] = df['cast'].apply(fetch_actors)
df['cast'].head()
```

###Crew

```python
df['crew'][0]
```

```python
def directors(x):
  D = []
  for d in ast.literal_eval(x):
    if d['department']=="Directing":
      D.append(d['name'])
  return D
```

```python
df['crew'] = df['crew'].apply(directors)
df['crew'].head()
```

```python
k = 0
for i in df.iloc[0]:
  print(f"{df.columns[k]}: {i} \n")
  k+=1
```

#Text Preprocessing

```python
df['genres'] = df['genres'].apply(lambda x:[i.replace(" ", "") for i in x ])
df['genres']
```

```python
df['keywords'] = df['keywords'].apply(lambda x:[i.replace(" ", "") for i in x ])
df['cast'] = df['cast'].apply(lambda x:[i.replace(" ", "") for i in x ])
df['crew'] = df['crew'].apply(lambda x:[i.replace(" ", "") for i in x ])
```

```python
df.head()
```

```python
df['tag'] = df['overview']+df['genres']+df['keywords']+df['cast']+df['crew']
df.iloc[0]['tag']
```

```python
import warnings
warnings.filterwarnings('ignore')

df['tag'] = df['tag'].apply(lambda x:" ".join(x))
df['tag'] = df['tag'].apply(lambda x:x.lower())
```

```python
final_df = df[['movie_id', 'title', 'tag']]
final_df.head()
```

```python
final_df['tag'][0]
```

```python
from sklearn.feature_extraction.text import CountVectorizer
cv = CountVectorizer(max_features=5000, stop_words='english')
```

```python
# !pip install nltk
```

```python
import nltk
from nltk.stem.porter import PorterStemmer
ps = PorterStemmer()
```

```python
def stem(txt):
  res = []
  for i in txt.split():
    res.append(ps.stem(i))
  return " ".join(res)
```

```python
final_df['tag'] = final_df['tag'].apply(stem)
```

#$Text -> Vectors$

using Bag of Words

```python
vecs = cv.fit_transform(final_df['tag']).toarray()
top_features = cv.get_feature_names_out()
for i in top_features:
  print(i)
```

```python
from sklearn.metrics.pairwise import cosine_similarity
similarity = cosine_similarity(vecs)
```

```python
similarity.shape
```

```python
similarity[0]
```

```python
def recommender(movie):
  idx = final_df[final_df['title']==movie].index[0]
  relation = similarity[idx]
  movies_closer = sorted(list(enumerate(relation)), reverse = True, key = lambda x:x[1])[1:6]
  for movie in movies_closer:
    print(final_df.iloc[movie[0]].title)
```

```python
recommender('Batman Begins')
```
